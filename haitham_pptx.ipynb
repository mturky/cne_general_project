```python
from pptx import Presentation

# Create a presentation object
prs = Presentation()

# Define a function to add a slide with title and bullet points
def add_bullet_slide(title, bullet_points):
    slide_layout = prs.slide_layouts[1]  # Title and Content layout
    slide = prs.slides.add_slide(slide_layout)
    title_placeholder = slide.shapes.title
    content_placeholder = slide.placeholders[1]

    title_placeholder.text = title
    content_placeholder.text = ''
    for point in bullet_points:
        p = content_placeholder.text_frame.add_paragraph()
        p.text = point
        p.level = 0

# Define a function to add a title slide
def add_title_slide(title, subtitle, presenter, date):
    slide_layout = prs.slide_layouts[0]  # Title Slide layout
    slide = prs.slides.add_slide(slide_layout)
    slide.shapes.title.text = title
    slide.placeholders[1].text = f"{subtitle}\n\nPresented by: {presenter}\nDate: {date}"

# Slides
add_title_slide(
    "Creating a New Customer Service Department Through Social Media",
    "Enhancing Customer Experience in the Digital Age",
    "[Your Name / Team Name]",
    "[Insert Date]"
)

slides_content = [
    ("Why Social Media for Customer Service?", [
        "4.7 billion people use social media worldwide.",
        "Customers expect real-time support.",
        "Public responses = brand visibility.",
        "Opportunity to build loyalty and trust."
    ]),
    ("Project Objectives", [
        "Create a dedicated team for social media support.",
        "Improve response time and satisfaction scores.",
        "Integrate with existing CRM tools.",
        "Establish KPIs for performance tracking."
    ]),
    ("Target Social Media Platforms", [
        "Facebook: High engagement, broad demographics",
        "Instagram: Visual queries, product support",
        "X (Twitter): Real-time response and crisis handling",
        "LinkedIn: B2B support",
        "WhatsApp / Messenger: Private, fast, personal support"
    ]),
    ("Proposed Team Structure", [
        "Social Media Manager",
        "Customer Service Agents (Shift-based)",
        "Escalation Specialist",
        "Analytics & Reporting Officer"
    ]),
    ("Tools for Social Media Customer Support", [
        "CRM Integration: Salesforce, Zendesk",
        "Social Listening Tools: Sprout Social, Hootsuite, Brandwatch",
        "Automation/AI: Chatbots, canned responses",
        "Analytics: Real-time dashboards, sentiment analysis"
    ]),
    ("Customer Query Workflow", [
        "Message Received → Auto-Acknowledge → Assign Agent → Respond →",
        "Escalate (if needed) → Close Ticket → Survey"
    ]),
    ("Key Performance Indicators", [
        "Response time (avg. <15 min)",
        "Resolution time",
        "CSAT (Customer Satisfaction Score)",
        "NPS (Net Promoter Score)",
        "Volume handled per agent",
        "Engagement rate"
    ]),
    ("Project Rollout Timeline", [
        "Month 1: Planning & recruitment",
        "Month 2: Training & tool setup",
        "Month 3: Soft launch",
        "Month 4: Full implementation",
        "Month 5+: Monitoring & optimization"
    ]),
    ("Anticipated Challenges", [
        "High volume → Use AI/chatbots for triage",
        "Negative comments → Set escalation policy",
        "24/7 coverage → Global shift planning",
        "Brand voice → Agent training"
    ]),
    ("Expected Impact", [
        "20% increase in CSAT",
        "Reduced email/call center load",
        "Increased brand loyalty",
        "Publicly visible service = marketing asset"
    ]),
    ("Let's Launch the Future of Customer Service", [
        "Get stakeholder buy-in",
        "Start pilot phase",
        "Align with marketing and IT teams",
        "Build for scale and feedback"
    ]),
    ("Questions & Discussion", []),
    ("Thank You", [
        "Let’s empower our customers where they already are."
    ])
]

for title, bullets in slides_content:
    add_bullet_slide(title, bullets)

# Save the file
prs.save("Customer_Service_Social_Media_Plan.pptx")
```
