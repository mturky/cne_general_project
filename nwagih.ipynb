```python
import pandas as pd
pd.set_option('display.max_columns', None)
```

```python
disc2022 = pd.read_excel(r"S:\Taro2y\Both 2022.xlsx", sheet_name='DISC.')
detail2022 = pd.read_excel(r"S:\Taro2y\Both 2022.xlsx", sheet_name='DETA.')
disc2023 = pd.read_excel(r"S:\Taro2y\Both 2023.xlsx", sheet_name='DISC.')
detail2023 = pd.read_excel(r"S:\Taro2y\Both 2023.xlsx", sheet_name='DETA.')

disc2024 = pd.read_excel(r"S:\Taro2y\Both 2024.xlsx", sheet_name='DISC.')
detail2024 = pd.read_excel(r"S:\Taro2y\Both 2024.xlsx", sheet_name='DETA.')
disc2025 = pd.read_excel(r"S:\Taro2y\Both 2025.xlsx", sheet_name='DISC.')
detail2025 = pd.read_excel(r"S:\Taro2y\Both 2025.xlsx", sheet_name='DETA.')
```

```python
disc2022['Action Date'] = pd.to_datetime(disc2022['Action Date'],dayfirst=True).dt.normalize()
detail2022['Effective Date'] = pd.to_datetime(detail2022['Effective Date'],dayfirst=True).dt.normalize()
both2022 = pd.merge(left=disc2022,right=detail2022, on='Customer Number')
both2022['date_diff'] = (both2022['Action Date'] -  both2022['Effective Date']).dt.days
both2022.to_csv('2022.csv', index=False)
```
