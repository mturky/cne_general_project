```python
import pandas as pd
pd.set_option('display.max_columns', None)
```

```python
pair = pd.read_excel(r"S:\01.09.25 Pairing.xlsx",dtype='str')
bein = pd.read_csv(r"S:\01.09.25\26418847_BEINDATANEWRPT.CSV",dtype=str)
```

```python
pair_copy = pair.copy()
pair_copy['DECODER'] = pair_copy['MESSAGE'].str.split(",").str[-1]
pair_copy['CREATED_DATE'] = pd.to_datetime(pair_copy['CREATED_DATE'])
na_filter = pair_copy['CUSTOMER_NUMBER'].isna()
sysadmin_filter = pair_copy['CREATED_BY_NAME']=='Super User'
invalid_filter = pair_copy['STATUS_NAME']=='Invalid'

pair_copy = pair_copy.loc[(~(na_filter & sysadmin_filter))]
pair_copy = pair_copy.loc[~(invalid_filter)]

pair_copy= pair_copy.sort_values(['CUSTOMER_NUMBER',"CREATED_DATE"],ascending=[True,False])
pair_copy = pair_copy.drop_duplicates(subset=['CUSTOMER_NUMBER'], keep="first")
```

```python
bein_copy = bein.copy()
active_filter = bein['Status']=='Active'
bein_copy=bein_copy.loc[active_filter]
bein_copy['Original Decoder'] = bein_copy['Decoder']
bein_copy['Decoder'] = bein_copy['Decoder'].str[1:]
bein_copy['Original Smart Card'] = bein_copy['Smart Card']

bein_copy['Smart Card'] = bein_copy['Smart Card'].str[:-1]
bein_copy = bein_copy.loc[:,['Customer Number','Original Smart Card','Smart Card','Original Decoder','Decoder']]
bein_copy = bein_copy.drop_duplicates()
```

```python
merged_dec = pd.merge(left=bein_copy,right=pair_copy, left_on='Decoder', right_on='DECODER', how='right')
merged_dec = merged_dec.loc[merged_dec['SMART_CARD']!=merged_dec['Smart Card']].drop_duplicates()
merged_dec = merged_dec.loc[~merged_dec['Decoder'].isna()]
# merged_dec = merged_dec.drop_duplicates()
```

```python
merged_sc = pd.merge(left=bein_copy,right=pair_copy, left_on='Smart Card', right_on='SMART_CARD', how='right')
merged_sc = merged_sc.loc[merged_sc['DECODER']!=merged_sc['Decoder']].drop_duplicates()
merged_sc = merged_sc.loc[~merged_sc['Decoder'].isna()]
# merged_sc = merged_sc.drop_duplicates()
```
