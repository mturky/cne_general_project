```python
import pandas as pd
from datetime import datetime , timedelta

yesterday = (datetime.now() - timedelta(days=1)).strftime('%Y-%m-%d')
```

```python
detailed = pd.read_csv(r"C:\Users\mturky\Downloads\30880509_DETAILNEWANDREN.CSV", dtype='str')
disconnection = pd.read_csv(r"C:\Users\mturky\Downloads\30880508_DailyDisconnectionReport.CSV", dtype='str')
```

```python
sub_types = ['beIN Quartar Installment'
             , 'CNE Subscriber'
             , 'MCE staff (CNE staff)'
             ,'BeIN sports CC'
             , 'beIN Bi Installment'
             , 'Corporate Subscriber'
             , 'Bein NC'
             , 'Bulk DTH customer'
             , 'beIN Installment Sub']
```

```python
detailed['Effective Date'] = pd.to_datetime(
    detailed['Effective Date'],
    dayfirst=True
).dt.strftime('%Y-%m-%d')


disconnection['ACTION_DATE_3'] = pd.to_datetime(
    disconnection['ACTION_DATE_3'],
    dayfirst=True
).dt.strftime('%Y-%m-%d')
```

```python
detailed['Event'].unique()
```

```python
detailed = detailed.loc[detailed['Effective Date'] == yesterday]

detailed = detailed.loc[(detailed['Package'].str.contains('prem',case=False))
                        |(detailed['Package'].str.contains('ulti',case=False))
                        |(detailed['Package'].str.contains('toget',case=False))]

detailed = detailed.loc[detailed['Customer Type'].isin(sub_types)]

detailed_renewal = detailed.loc[detailed['Event'].isin(['New Agreement For Existing Customer', 'Renewal'])]
detailed = detailed.loc[detailed['Event'].isin(['New Agreement For Existing Customer', 'Activated from Suspension'])]
```

```python
new_for_existing = detailed.loc[detailed['Event']=='New Agreement For Existing Customer']
activated_from_suspension = detailed.loc[detailed['Event']=='Activated from Suspension']
```

```python
disconnection = disconnection.loc[(disconnection['AFFECTED_PLAN_5'].str.contains('prem',case=False))
                        |(disconnection['AFFECTED_PLAN_5'].str.contains('ulti',case=False))
                        |(disconnection['AFFECTED_PLAN_5'].str.contains('toget',case=False))]
```

```python
disconnection = disconnection.drop_duplicates(subset=['CUSTOMER_NUMBER_1'])
```

```python
new_none_common = new_for_existing.loc[~new_for_existing['Customer Number'].isin(disconnection['CUSTOMER_NUMBER_1'])]
```

```python
common_with_renewal = detailed_renewal.loc[detailed_renewal['Customer Number'].isin(disconnection['CUSTOMER_NUMBER_1'])]
common_with_renewal
```

```python
total_reinstatment = pd.concat([new_none_common,activated_from_suspension])
total_reinstatment
```
