```python
arr = [
    [1,2,3],
    [4,5,6],
    [7,7,9]
]
```

```python
def printArray(arr):
    for row in arr:
        for element in row:
            print(element,end=" ")
        print()
```

```python
arr[2][1] = 8
```

```python
printArray(arr)
```

```python
from dataclasses import dataclass
class Car:
    model:str
    color:str
    year:str

    def __init__(self,model,color,year):
        self.model = model
        self.color = color
        self.year = year
    
    def __repr__(self):
        return f"{self.model}, {self.color},{self.year}"



@dataclass
class Person:
    name:str
    age:int
    car:Car
```

```python
car1 = Car("BMW","YELLOW","1980")
car2 = Car("NISSAN","GREEN","1814")

omar = Person("Omar Ahmed",17,car2)

print(car1)
```

```python
omar.car = car2
print(omar,omar.car)
```
