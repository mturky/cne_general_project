```python
import pandas as pd
```

```python
data = pd.read_csv(r'S:\17.07.25\25712180_BEINDATANEWRPT.CSV', dtype='str')
data=data.loc[data['Contract Number']!='2098194']
```

```python
data.loc[data['Status']=='Active','Status']='1 Active'
data.loc[data['Status']=='Suspended','Status']='2 Suspended'

data.loc[data['Status']=='DIS','Status']='3 Disconnected'

```

```python
data1000s = data.loc[data['Item Description STB']=='beIN Decoder 1000s'].copy()
data1000s['End Date'] = pd.to_datetime(data1000s['End Date'],dayfirst=True)
```

```python
data1000s['payment_plan'] = 'Non Installment'
data1000s.loc[data1000s['Customer Type'].str.contains('Installment'),'payment_plan'] = 'Installment'
# data1000s.loc[data1000s['Status']=='Suspended','Status']='DIS'
```

```python
data1000s.columns
```

```python
data1000s = data1000s.sort_values(['Status','End Date'], ascending=[True,False])
data1000s= data1000s.drop_duplicates(subset=['Decoder'], keep='first')
data1000s= data1000s.reset_index()
data1000s['month'] = data1000s['End Date'].dt.month
data1000s['year'] = data1000s['End Date'].dt.year
```

```python
data1000s.loc[data1000s['Status']=='1 Active','Status']='Active'
data1000s.loc[data1000s['Status']=='2 Suspended','Status']='Suspended'

data1000s.loc[data1000s['Status']=='3 Disconnected','Status']='Disconnected'

data1000s.to_csv('data1000s_clean.csv', index = False)
```
