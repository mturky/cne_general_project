```python
import pandas as pd
pd.set_option('display.max_columns', None)
```

```python
data = pd.read_excel(r"C:\Users\mturky\Documents\Commercial_Test_Jan2021-Jul2025.xlsx",dtype='str')
```

```python
data['Created Date'] = pd.to_datetime(data['Created Date'])
```

```python
data['month'] = data['Created Date'].dt.month
data['year'] = data['Created Date'].dt.year
```

```python
data2025 = data.loc[data['Created Date']>='2025-01-01']
data_before = data.loc[data['Created Date']<'2025-01-01']
```

```python
renewal = data2025.loc[data2025['Subscriber Nr'].isin(data_before['Subscriber Nr'])]
new = data2025.loc[~data2025['Subscriber Nr'].isin(data_before['Subscriber Nr'])]

renewal.to_excel('ihab_commercial_renewal.xlsx',index= False)
new.to_excel('ihab_commercial_new.xlsx',index= False)

```

```python
new.groupby(['year','month']).agg(count=('Subscriber Nr','nunique')).reset_index().to_excel("ihab_commercial_new_totals.xlsx")
```

```python
renewal.groupby(['year','month']).agg(count=('Subscriber Nr','nunique')).reset_index().to_excel("ihab_commercial_renewal_totals.xlsx")
```
