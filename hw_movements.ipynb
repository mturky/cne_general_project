```python
import pandas as pd
import numpy as np
```

```python
swap_original = pd.read_csv(r"S:\Sheikh\30592658_SWAPRPT.CSV",dtype='str')
bein_original = pd.read_csv(r"S:\21.07.26\30590257_BEINDATANEWRPT.CSV",dtype='str')
```

```python
bein = (bein_original[['Customer Number','Smart Card', 'Item Description SC','Decoder',
       'Item Description STB']]
       .drop_duplicates().copy())

swap = swap_original[['Swap Date','Subscriber Number','Item', 'Old Serial Number','New Serial Number']].drop_duplicates().copy()
swap[['Old SC', 'New SC', 'Old STB', 'New STB']] = ''
swap = swap.loc[swap['Item']!='ART Smart Card']
```

```python

swap.loc[
    swap['Item'] == 'Smartcard',
    ['Old SC', 'New SC']
] = swap.loc[
    swap['Item'] == 'Smartcard',
    ['Old Serial Number', 'New Serial Number']
].to_numpy()

swap.loc[
    swap['Item'] == 'Decoder',
    ['Old STB', 'New STB']
] = swap.loc[
    swap['Item'] == 'Decoder',
    ['Old Serial Number', 'New Serial Number']
].to_numpy()

swap = swap[['Swap Date', 'Subscriber Number', 'Item',  'Old SC', 'New SC', 'Old STB', 'New STB']]
swap['Swap Date'] = pd.to_datetime(swap['Swap Date'],dayfirst=True).dt.normalize()

swap = swap.replace('',np.nan)
```
