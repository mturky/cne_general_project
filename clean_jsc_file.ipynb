```python
import pandas as pd
from functions import search_files
import csv
```

```python
ft = pd.read_csv(search_files("NEWCNEFIN")[0], dtype='str', nrows=0)

rows = []

with open(search_files("NEWCNEFIN")[0], encoding="utf-8", newline="") as f:
    reader = csv.reader(f)

    header = next(reader)

    # Keep columns except problematic column
    header = header[:13] + header[-19:]
    rows.append(header)

    for row in reader:
        if len(row) >= 10:
            # Keep first 3 + last 6
            row = row[:13] + row[-19:]
            rows.append(row)

df = pd.DataFrame(rows[1:], columns=rows[0])
print(df.columns)
df.to_csv('clean ft.csv', index=False)
```
