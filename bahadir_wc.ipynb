```python
import pandas as pd
pd.set_option('display.max_columns', None)
```

```python
file = pd.read_csv(r"S:\04.06.26\30301307_NEWCNEFINTRANSRPT.CSV", on_bad_lines='skip')
```

```python
SUB_TYPES = [
        'beIN Quartar Installment', 'CNE Subscriber', 'MCE staff (CNE staff)',
        'BeIN sports CC', 'beIN Bi Installment', 'Corporate Subscriber', 'Temp',
        'Bein NC',  'beIN Installment Sub', 'Charge Back',
        'Illegal Network',
        'Head End','Outsiders'
    ]

file = file.loc[file['Subscriber Type'].isin(SUB_TYPES)]
```

```python
file['Created Date'] = pd.to_datetime(file['Created Date'], dayfirst=  True)
file['Period From'] = pd.to_datetime(file['Period From'], dayfirst=  True)
```

```python
data = file.loc[file['Created Date']>='2026-05-25']
data = file.loc[file['Period From']>='2026-05-25']
```

```python
data = data.loc[data['Doc Type'].isin(['Invoice'])]
data = data.loc[data['Doc Status'].isin(['Posted'])]
data['p'] = data['Plan Name']
```

```python
data.loc[data['Plan Name'].str.startswith('FWC 2026'),'p'] = "Summer Offer"
data.loc[data['Plan Name'].str.startswith('ULTIMATE'),'p'] = "Ultimate"
```

```python

plans=['ULTIMATE 6 Months Contract','ULTIMATE 09.24','Ultimate 2024','ULTIMATE 6 Months Waiver','Ultimate','NCULTIMATE 09.24','FWC 2026- Together 12M Complmtry',
      'FWC 2026- Premium 3M Complmtry','AddON FWC 2026','SA FWC 2026', ]
```

```python
data = data.loc[data['Plan Name'].isin(plans)]
```

```python
data.to_csv('bahadir.csv',index= False)
```
