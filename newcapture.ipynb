```python
import pandas as pd
from datetime import datetime , timedelta

yesterday = (datetime.now() - timedelta(days=1)).strftime('%Y-%m-%d')
sub_types = ['beIN Quartar Installment', 'CNE Subscriber', 'MCE staff (CNE staff)',
            'BeIN sports CC', 'beIN Bi Installment', 'Corporate Subscriber', 
            'Bein NC', 'Bulk DTH customer', 'beIN Installment Sub']
```

```python
new_capture = pd.read_csv(r"S:\07.09.26\30879285_CNENEWCAPTURERPT.CSV", dtype='str')

new_capture['Start Date'] = pd.to_datetime(
    new_capture['Start Date'],
    dayfirst=True
).dt.strftime('%Y-%m-%d')
```

```python
new_capture = new_capture.loc[new_capture['Customer Type'].isin(sub_types)]
```

```python
new_capture = new_capture.loc[new_capture['Start Date']== yesterday]
new_capture = new_capture.loc[(new_capture['Plan'].str.contains('prem',case=False))
                        |(new_capture['Plan'].str.contains('ulti',case=False))
                        |(new_capture['Plan'].str.contains('toget',case=False))]
```
