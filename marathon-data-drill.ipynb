```python
import pandas as pd
```

```python
df = pd.read_csv(r"C:\Users\mturky\Downloads\marathon-data.csv")
```

```python
df['final_minutes'] = pd.to_timedelta(df['final']).dt.total_seconds() / 60
```

```python
bands = [0,180,210,240,270,300,330,360,float('inf')]
bands_labels = ["sub 3", "3-3:30","3:30-4","4-4:30","4:30-5","5-5:30","5:30-6","6+"]
df["bands"] = pd.cut(df['final_minutes'],bands,labels=bands_labels,right=False)
```

```python
df.bands.value_counts(normalize=True)
```
