```python
from pathlib import Path
import pandas as pd

# Folder containing CSV files
folder = Path(r"C:\Users\mturky\Documents\orange data\15-7-2026")

# Get all CSV files
csv_files = folder.glob("*.csv")

# Read and concatenate
df = pd.concat(
    (pd.read_csv(file) for file in csv_files),
    ignore_index=True
)

print(df.head())
print(df.shape)

df.to_csv('df15072026.csv', index=False)
```
