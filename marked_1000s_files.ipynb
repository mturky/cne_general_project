```python
import pandas as pd
pd.set_option("display.max_columns",None)
```

```python
swap = pd.read_csv(r"S:\1000s project\1000s for swap.csv",dtype='str', on_bad_lines='skip')
bein = pd.read_csv(r"S:\29.09.25\26757323_BEINDATANEWRPT.CSV",dtype='str')
```

```python
bein_active = bein.loc[(bein['Status']=='Active')]
bein_active= bein_active[['Customer Number','Customer Type','Smart Card','Decoder','Item Description STB']]
```

```python
bein_dec_sc = bein[['Smart Card','Decoder','Item Description STB']]
```

```python
active = swap.loc[swap['Customer Number'].isin(bein_active['Customer Number'])]
```

```python
result = active.merge(bein_dec_sc, on=['Decoder'],how='left').drop_duplicates(subset='Customer Number')
```

```python
result = result[['Customer Number','Customer Type','Item Description STB']]
result = result.rename({"Item Description STB":"status"},axis=1)
result.loc[result['status'].isna(),'status'] = 'Swapped'
result.loc[result['status']!='Swapped','status'] = 'Renewed'
result.to_csv("")
```
