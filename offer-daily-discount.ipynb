```python



start_date = datetime.strptime('2025-06-03', '%Y-%m-%d')
number_of_days = 50


plans =[

    {'plan':"Premium", "payment_plan":24,"initial_price":12168},
    {'plan':"Together", "payment_plan":24,"initial_price":5880},

    {'plan':"Premium", "payment_plan":12,"initial_price":6087},
    {'plan':"Together", "payment_plan":12,"initial_price":2940},

    # {'plan':"Premium", "payment_plan":6,"initial_price":3045},
    # {'plan':"Together", "payment_plan":6,"initial_price":1470},

    # {'plan':"Premium", "payment_plan":3,"initial_price":1522},
    # {'plan':"Together", "payment_plan":3,"initial_price":735}
    ]
```

```python
rows = []
for item in plans:
    for i in range(number_of_days):
        new_price = item['initial_price'] - (item['initial_price'] * ((50 - i)) / 100)
        new_price_rounded = round(new_price)
        discount_rate = 50-i
        new_date = start_date + pd.DateOffset(days=i)
        row = {
            'Date': new_date,
              'Payment Plan': item['payment_plan'],
              'Package': item["plan"],
                'Discount': f'{discount_rate}%' ,
                'Offer Price (With VAT)': new_price_rounded,
                  'Offer Package Name' : f'{str(item["plan"]).upper()}2025SUMMER{discount_rate}',
                  'System_plan': f'{str(item["plan"])} 2025 SUMMER {discount_rate}'}
        rows.append(row)


df = pd.DataFrame(rows)
df.to_excel("result.xlsx", sheet_name='Sheet1', index=False)
```
