```python
import pandas as pd
```

```python
file1 = r"S:\playground\commercial.csv"
file2 = r"S:\playground\commercial.csv"

df1 = pd.read_csv(file1,dtype='str', on_bad_lines='skip')
df2 = pd.read_csv(file2,dtype='str', on_bad_lines='skip')

field1 = 'Smart Card'
field2 = 'Smart Card'

in1notin2 = df1.loc[~df1[field1].isin(df2[field2])]
in2notin1 = df2.loc[~df2[field2].isin(df1[field1])]
both = df1.loc[df1[field1].isin(df2[field2])]

in1notin2.to_csv('in 1 not in 2.csv',index=False)
in2notin1.to_csv('in 2 not in 1.csv',index=False)
both.to_csv('in both files.csv',index=False)
```
