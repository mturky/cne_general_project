```python
import pandas as pd
pd.set_option('display.max_columns', None)
```

```python
all = pd.read_csv(r"C:\Users\mturky\Downloads\26746385_ALLUSERHISTDTLS\26746385_ALLUSERHISTDTLS.CSV",dtype='str')
modify = pd.read_csv(r"C:\Users\mturky\Downloads\26746386_MODIFYSUBSTYPERPT.CSV",dtype='str')
```

```python
filter_sharaf = modify['User Code'].str.lower().str.contains('sharaf')
filter_cne = modify['User Code'].str.lower().str.contains('cne')
filter_edd = modify['User Code'].str.lower().str.contains('edd')
filter_dis = modify['User Code'].str.lower().str.contains('dis')
filter_bein = modify['User Code'].str.lower().str.contains('bein')
```

```python
all_copy = all.loc[all['Transaction Type']=="Post Paid"].copy()
modify_copy = modify.loc[filter_cne | filter_edd | filter_dis | filter_sharaf | filter_bein].copy()
```

```python
all_copy["Transaction Date"] = pd.to_datetime(all_copy["Transaction Date"],dayfirst=True)
modify_copy['Modified Date'] = pd.to_datetime(modify_copy['Modified Date'],dayfirst=True)
all_copy['Transaction Date'] = pd.to_datetime(all_copy['Transaction Date'].dt.date, dayfirst=True)
```

```python
merged = pd.merge(left=modify_copy,right=all_copy,left_on=['Customer Number','Modified Date'] , right_on=['Customer#','Transaction Date'], how="left")
```

```python
result = merged.loc[merged['Customer#'].isna()]
result = result.drop_duplicates(subset='Customer Number')
```

```python
result.to_csv("sheikh result.csv", index=False)
```
