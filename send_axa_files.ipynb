```python
import os
import pandas as pd

from functions import sendMail

def send_folders_by_email(base_folder, csv_file):
    employees = pd.read_csv(csv_file)

    for _, row in employees.iterrows():
        employee_id = str(row["ID"]).strip()
        email = str(row["Email"]).strip()

        if not employee_id or not email or email.lower() == "nan":
            continue

        # Folder corresponding to employee ID
        folder = os.path.join(base_folder, employee_id)

        if not os.path.isdir(folder):
            print(f"Folder not found: {employee_id}")
            continue

        # Get all files inside the folder
        attachments = []

        for filename in os.listdir(folder):
            filepath = os.path.join(folder, filename)

            if os.path.isfile(filepath):
                attachments.append(filepath)

        if not attachments:
            print(f"No files found in folder: {employee_id}")
            continue

        subject = f"Medical Cards AXA 2026-2027 - {employee_id}"



        print(f"Sending {employee_id} -> {email}")
        print(f"Attachments: {len(attachments)}")

        sendMail(
            subject=subject,
            body=body,
            to=[email],
            attachments=attachments
        )

        print(f"Sent successfully: {employee_id}\n")
```
